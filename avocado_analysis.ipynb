```python
import pandas as pd
import numpy as np
import scipy.stats as st
from scipy.stats import t
from scipy.stats import norm
import matplotlib.pyplot as plt
from matplotlib.dates import DateFormatter

avocado = pd.read_csv("avocado.csv")

# Mean of Total Volume avocado sold weekly in TOTAL US from 2015 to 2018
totalus_rows = avocado[avocado["region"] == "TotalUS"]
totalus_mean = np.mean(totalus_rows["Total Volume"])
print("The mean of the Total Volume of avocados sold in TOTAL US is:", round(totalus_mean))

# Mean of PLU 4046 avocado sold weekly in TOTAL US from 2015 to 2018
totalus_rows = avocado[avocado["region"] == "TotalUS"]
totalus_mean = np.mean(totalus_rows["4046"])
print("The mean of PLU 4046 avocados sold in TOTAL US is:", round(totalus_mean))

# Mean of PLU 4770 avocado sold weekly in TOTAL US from 2015 to 2018
totalus_rows = avocado[avocado["region"] == "TotalUS"]
totalus_mean = np.mean(totalus_rows["4770"])
print("The mean of PLU 4770 avocados sold in TOTAL US is:", round(totalus_mean))

# Mean of PLU 4225 avocado sold weekly in TOTAL US from 2015 to 2018
totalus_rows = avocado[avocado["region"] == "TotalUS"]
totalus_mean = np.mean(totalus_rows["4225"])
print("The mean of PLU 4225 avocados sold in TOTAL US is:", round(totalus_mean))

# Mean of Total Bags volume sold weekly in TOTAL US from 2015 to 2018
totalus_rows = avocado[avocado["region"] == "TotalUS"]
totalus_mean = np.mean(totalus_rows["Total Bags"])
print("The mean of Total Bags volume sold in TOTAL US is:", round(totalus_mean))

# Mean of Avergae Price avocado sold weekly in TOTAL US from 2015 to 2018
totalus_rows = avocado[avocado["region"] == "TotalUS"]
totalus_mean = np.mean(totalus_rows["AveragePrice"])
totalus_mean_rounded = round(totalus_mean, 3)
print("The mean of Average Price avocados sold in TOTAL US is:", totalus_mean_rounded)

# Graph 1 (monthly mean of PLU 4046 avocado sold in TotalUS during 2015)
filtered_data = avocado.loc[(avocado['region'] == 'TotalUS') & (avocado["year"] == 2015) & (avocado["type"] == "conventional"), :].copy()
filtered_data["Date"] = pd.to_datetime(filtered_data["Date"])
filtered_data["Month"] = filtered_data["Date"].dt.month
monthly_data = filtered_data.groupby("Month")["4046"].mean()

# Create the graph 1
plt.plot(monthly_data.index, monthly_data.values)
plt.xlabel('Month')
plt.ylabel('Volume 4046')
plt.title('conventional PLU 4046 sold during 2015')
plt.xticks(rotation=45)
plt.show()

# Graph 2 (monthly mean of PLU 4046 avocado sold in TotalUS during 2016)
filtered_data = avocado.loc[(avocado['region'] == 'TotalUS') & (avocado["year"] == 2016) & (avocado["type"] == "conventional"), :].copy()
filtered_data["Date"] = pd.to_datetime(filtered_data["Date"])
filtered_data["Month"] = filtered_data["Date"].dt.month
monthly_data = filtered_data.groupby("Month")["4046"].mean()

# Create the graph 2
plt.plot(monthly_data.index, monthly_data.values)
plt.xlabel('Month')
plt.ylabel('Volume 4046')
plt.title('conventional PLU 4046 sold during 2016')
plt.xticks(rotation=45)
plt.show()

# Graph 3 (monthly mean of Total Bags avocado sold in TotalUS during 2015)
filtered_data = avocado.loc[(avocado['region'] == 'TotalUS') & (avocado["year"] == 2015) & (avocado["type"] == "conventional"), :].copy()
filtered_data["Date"] = pd.to_datetime(filtered_data["Date"])
filtered_data["Month"] = filtered_data["Date"].dt.month
monthly_data = filtered_data.groupby("Month")["Total Bags"].mean()

# Create the graph 3
plt.plot(monthly_data.index, monthly_data.values)
plt.xlabel('Month')
plt.ylabel('Total Bags')
plt.title('coventional Total Bags sold during 2015')
plt.xticks(rotation=45)
plt.show()

# Graph 4 (monthly mean of Total Bags avocado sold in TotalUS during 2016)
filtered_data = avocado.loc[(avocado['region'] == 'TotalUS') & (avocado["year"] == 2016) & (avocado["type"] == "conventional"), :].copy()
filtered_data["Date"] = pd.to_datetime(filtered_data["Date"])
filtered_data["Month"] = filtered_data["Date"].dt.month
monthly_data = filtered_data.groupby("Month")["Total Bags"].mean()

# Create the graph 4
plt.plot(monthly_data.index, monthly_data.values)
plt.xlabel('Month')
plt.ylabel('Total Bags')
plt.title('coventional Total Bags sold during 2016')
plt.xticks(rotation=45)
plt.show()


#PART 2 QUESTION 1

#monthly mean of total bags sold in Los Angeles during 2015
filtered_data = avocado.loc[(avocado['region'] == 'LosAngeles') & (avocado["year"] == 2015) & (avocado["type"] == "conventional"), :].copy()
filtered_data["Date"] = pd.to_datetime(filtered_data["Date"])
filtered_data["Month"] = filtered_data["Date"].dt.month
monthly_data1 = filtered_data.groupby("Month")["Total Bags"].mean()

#graph 2015
plt.plot(monthly_data1.index, monthly_data1.values)
plt.xlabel('Month')
plt.ylabel('Total Bags Volume')
plt.title('Monthly Mean of Total Bags Sold in Los Angeles during 2015')
plt.xticks(rotation=45)
plt.show()

#logarithm of monthly mean of total bags sold in Los Angeles during 2015
log_transformed_data1 = np.log(monthly_data1)
plt.plot(monthly_data1.index, log_transformed_data1)
plt.xlabel('Month')
plt.ylabel('Log-Transformed Total Bags Volume')
plt.title('Monthly Mean of Total Bags Sold in Los Angeles during 2015 (Log-Transformed)')
plt.xticks(rotation=45)
plt.show()

#monthly mean of total bags sold in Los Angeles during 2016
filtered_data = avocado.loc[(avocado['region'] == 'LosAngeles') & (avocado["year"] == 2016) & (avocado["type"] == "conventional"), :].copy()
filtered_data["Date"] = pd.to_datetime(filtered_data["Date"])
filtered_data["Month"] = filtered_data["Date"].dt.month
monthly_data2 = filtered_data.groupby("Month")["Total Bags"].mean()

#graph 2016
plt.plot(monthly_data2.index, monthly_data2.values)
plt.xlabel('Month')
plt.ylabel('Total Bags Volume')
plt.title('Monthly Mean of Total Bags Sold in Los Angeles during 2016')
plt.xticks(rotation=45)
plt.show()

#logarithm of monthly mean of total bags sold in Los Angeles during 2016
log_transformed_data2 = np.log(monthly_data2)
plt.plot(monthly_data2.index, log_transformed_data2)
plt.xlabel('Month')
plt.ylabel('Log-Transformed Total Bags Volume')
plt.title('Monthly Mean of Total Bags Sold in Los Angeles during 2016 (Log-Transformed)')
plt.xticks(rotation=45)
plt.show()

#paired t test
filtered_data1 = avocado.loc[(avocado['region'] == 'LosAngeles') & (avocado["year"] == 2015) & (avocado["type"] == "conventional"), :].copy()
filtered_data2 = avocado.loc[(avocado['region'] == 'LosAngeles') & (avocado["year"] == 2016) & (avocado["type"] == "conventional"), :].copy()

# Select the first 40 rows for each year
subset_2015 = filtered_data1.head(40).reset_index(drop=True)
subset_2016 = filtered_data2.head(40).reset_index(drop=True)

bags_2015 = subset_2015['Total Bags']
bags_2016 = subset_2016['Total Bags']
difference = bags_2015 - bags_2016

# Apply log transformation
log_bags_2015 = np.log(bags_2015)
log_bags_2016 = np.log(bags_2016)
log_difference = log_bags_2015 - log_bags_2016

# Create a new DataFrame with the desired columns
mini_dataset = pd.DataFrame({'log total bags 2015': log_bags_2015, 'log total bags 2016': log_bags_2016, 'log difference': log_difference})
# Print the mini dataset
display(mini_dataset)

#t statistic
mean_difference = log_difference.mean()
std_difference = log_difference.std()
sample_size = 40
degrees_of_freedom = sample_size - 1
std_error = std_difference / np.sqrt(sample_size)
t_statistic = mean_difference / std_error
print("T statistic is: ", t_statistic)
alpha = 0.1  # 1 - confidence level (90% confidence interval)
# Find the critical value
critical_value = t.ppf(1 - alpha/2, degrees_of_freedom)
# Perform the t-test
if abs(t_statistic) > critical_value:
    print("Reject the null hypothesis for question 1")
else:
    print("Fail to reject the null hypothesis for question 1")
    
#PART 2 QUESTION 2 (Z test)
chicago_data = avocado.loc[(avocado['region'] == 'Chicago') & (avocado["year"] == 2015) & (avocado["type"] == "conventional"), :].copy()

# Select the columns of interest
plu_4046_column = chicago_data['4046']
total_bags_column = chicago_data['Total Bags']

# Calculate the variances
variance_plu_4046 = plu_4046_column.var()
variance_total_bags = total_bags_column.var()

print("Variance for PLU 4046 column:", variance_plu_4046)
print("Variance for Total Bags column:", variance_total_bags)

# Calculate the means
plu_4046_column_sample = plu_4046_column.head(40)
total_bags_column_sample = total_bags_column.head(40)

avg_plu_4046 = plu_4046_column_sample.mean()
avg_total_bags = total_bags_column_sample.mean()
print("The PLU 4046 sample average is: ", avg_plu_4046)
print("The total bags sample average is: ", avg_total_bags)

n1 = n2 = 40
d0 = 0
alpha2 = 0.05

Zd = (avg_plu_4046 - avg_total_bags -d0)/np.sqrt((variance_plu_4046/n1) + (variance_total_bags/n2))
print("The Z statistic is: ", Zd)
critical_value2 = norm.ppf(alpha2)
print("The critical value is: ", critical_value2)
if Zd < critical_value2:
    print("Reject the null hypothesis for question 2")
else:
    print("Fail to reject the null hypothesis for question 2")
    
```
